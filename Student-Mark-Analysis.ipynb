# Student Marks Analysis

## Description

This project analyzes student marks and calculates the total,
average, highest, and lowest marks.

## Objective

The objective is to practice Python data processing and
basic statistical calculations.

```python
marks = [85, 72, 90, 65, 78, 88, 95, 60, 82, 75]
total = sum(marks)

average = total / len(marks)

highest = max(marks)
lowest = min(marks)


print("Student Marks Analysis")
print("----------------------")
print("Marks:", marks)
print("Total Marks:", total)
print("Average Marks:", average)
print("Highest Mark:", highest)
print("Lowest Mark:", lowest)
```
